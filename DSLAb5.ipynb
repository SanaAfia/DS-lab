```python
import matplotlib.pyplot as plt
months = ["jan","feb","mar","apr","jun"]
sales = [120,130,140,150,160]
expenses = [80,90,100,120,130]
fig, ax = plt.subplots(2, 1, figsize=(10,5))
ax[0].plot(
    months,
    sales,
    marker="o",
    linewidth=2,
    label="sales"
)
ax[0].set_title("monthly sales")
ax[0].set_xlabel("month")
ax[0].set_ylabel("sales")
ax[0].set_xticks(range(len(months)))
ax[0].set_xticklabels(months)
ax[0].annotate(
    "Highest sales",
    xy=(5,250),
    xytext=(2,300),
    arrowprops=dict(arrowstyle="->")
)
ax[0].legend()
ax[1].plot(
    months,
    expenses,
    marker="o",
    linewidth=2,
    label="expenses"
)
ax[1].set_title("montly expenses")
ax[1].set_xlabel("month")
ax[1].set_ylabel("expenses")
ax[1].set_xticks(range(len(months)))
ax[1].set_xticklabels(months)
ax[1].annotate(
    "highest expenses",
    xy=(5,150),
    xytext=(2,200),
    arrowprops=dict(arrowstyle="->")
)
ax[1].legend()
ax[1].grid(True)
plt.tight_layout()
plt.savefig(
    "monthly_sales_expenses.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    'python': [30,40,50,60],
    'java' : [75,85,95,45],
}
df=pd.DataFrame(
    data,
    index=['stu1','stu2','stu3','stu4']
)
print("DataFrame:")
print(df)
df.plot(
    kind='bar',
    figsize=(8,9)
)
plt.title("student marks in diff programming langauges")
plt.xlabel("students")
plt.ylabel("marks")
plt.xticks(rotation=0)
plt.legend(title="subjects")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
data ={
    'python':[30,40,50,60],
    'java':[75,85,95,45]
}
df = pd.DataFrame(
    data,
    index=['stu1','stu2','stu3','stu4']
)
print("DataFrame:")
print(df)
df.plot(
    kind='bar',
    stacked=True,
    figsize=(8,5)
)
plt.title("student bar plot of student marks")
plt.xlabel("students")
plt.ylabel("total marks")
plt.xticks(rotation=0)
plt.legend(title="subjects")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data =[45,46,47,48,50,60,70,80,90]
df = pd.DataFrame({'marks': data})
print("Oberseved data:")
print(df)
plt.title("Histogram of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
plt.figure(figsize=(8,5))
sns.kdeplot(df['marks'],fill=True)
plt.title("Density plot of Marks")
plt.xlabel("marks")
plt.ylabel("density")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
study_hours=[1,2,3,4,5,6,7,8]
marks=[45,44,50,39,49,33,46,47]
df=pd.DataFrame(
    {'study_hours':study_hours,
    'marks':marks}
)
print("Observed data:")
print(df)
plt.figure(figsize=(8,5))
plt.scatter(df['study_hours'], df['marks'])
plt.title("Realationship between study hours and marks")
plt.xlabel("study hours")
plt.ylabel("marks")
plt.grid(True)
plt.tight_layout()
plt.show()
correlation = df['study_hours'].corr(df['marks'])
print("\ncorrelation coefficient:",round(correlation, 2))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = {
    'department':[
        'cse','ds','it','ds','ds'
    ],
    'marks':[
        90,88,75,80,95
    ]
}
df=pd.DataFrame(data)
print("Data:")
print(df)
plt.figure(figsize=(8,5))
sns.boxplot(x='department',y='marks',data=df)
plt.title("Boxplot of marks by department")
plt.xlabel("department")
plt.ylabel("marks")
plt.tight_layout()
plt.show()
```
