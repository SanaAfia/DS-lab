```python
import sqlite3
import pandas as pd
# connect to database
conn = sqlite3. connect("students.db")
cursor = conn.cursor()
# Create table
cursor.execute(
    "CREATE TABLE IF NOT EXISTS students(id INTEGER,name TEXT)"
)
# Insert data
cursor.execute("INSERT INTO students VALUES(1,'Sana')")

cursor.execute("INSERT INTO students VALUES(2,'Jhansi')")
conn.commit()
# Read data
df = pd.read_sql_query("SELECT *FROM students",conn)
print(df)
# Update and delete
cursor.execute(
    "UPDATE students SET name = 'Afia' WHERE id=1"
)
cursor.execute(
    "DELETE FROM students WHERE id = 1"
)
conn.commit()
conn.close()
```
