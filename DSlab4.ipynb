```python
import pandas as pd
index = [
    ['Engineering','Engineering','Science','Science'],
    ['cse','Ds','phy','chem']
]
multi_index = pd.MultiIndex.from_arrays(
    index,
    names = ['Department','Branch'] #multiindex
)
marks = pd.Series(
    [85,78,92,88],
    index = multi_index
)
print("original series: ")
print(marks)
#partial indexing
print("\nData for Engineering:")
print(marks.loc['Engineering'])
print("\nData for cse:")
print(marks.loc[('Engineering','cse')])
print("\nData for Science:")
print(marks.loc['Science'])
```

```python
import pandas as pd
index = pd.MultiIndex.from_tuples(
    [
        ('Engineering','cse'),
        ('Engineering','Ds'),
        ('Science','phy'),
        ('Science','chem')
    ],
    names = ['Department','Branch']
)
data = pd.DataFrame(
    {
        '2025':[85,78,92,88],
        '2026':[90,82,95,91]
    },
    index = index
)
print("original dataframe: ")
print(data)
unstacked_data = data.unstack()
print("\nunstacked dataframe: ")
print(unstacked_data)
stacked_data = unstacked_data.stack()
print("\nstacked dataframe: ")
print(stacked_data)
```
